```python
# Read source
from pyspark.sql import functions as F

orders=[
    (101,1,"2026-09-20",500.0),
    (102,1,"2026-09-21",1200.0),
    (103,2,"2026-09-21",800.0),
    (104,3,"2026-09-22",2500.0),
    (105,2,"2026-09-23",1500.0)
]

columns=["order_id","customer_id","order_date","amount"]

orders_df = spark.createDataFrame(orders,columns)

orders_df = orders_df.withColumn(
    "order_date",
    F.to_date("order_date")
)

display(orders_df)
```

```python
bronze_path= "/Volumes/workspace/default/filestore/day12/bronze/orders"

orders_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(bronze_path)

print("Bronze ingestion completed")
```

```python
bronze_df = spark.read.format("delta").load(bronze_path)

print("Bronze record count:", bronze_df.count())

display(bronze_df)
```

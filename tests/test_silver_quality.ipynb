```python
passed = 0
total = 0
```

# Test-1 Negative amounts

```python
from pyspark.sql import functions as F

test_orders = [
    (101,1,"2026-09-20",500.0),
    (102,1,"2026-09-21",-50.0),
    (103,2,"2026-09-22",800.0)
]

columns = ["order_id","customer_id","order_date","amount"]

test_df = spark.createDataFrame(test_orders,columns)
```

```python
valid_df = test_df.filter(
    (F.col("order_id").isNotNull()) &
    (F.col("customer_id").isNotNull()) &
    (F.col("order_date").isNotNull()) &
    (F.col("amount").isNotNull()) &
    (F.col("amount")>=0)
)
```

```python
total+=1
assert valid_df.count() == 2 , "Negative amount was not rejected"

passed+=1
print("Test 1 passed")
```

# TEST - 2


```python
test_orders = [
    (101, 1, "2026-09-20", 500.0),
    (102, None, "2026-09-21", 1200.0),
    (103, 2, "2026-09-22", 800.0)
]

test_df = spark.createDataFrame(test_orders, columns)

valid_df = test_df.filter(
    (F.col("order_id").isNotNull()) &
    (F.col("customer_id").isNotNull()) &
    (F.col("order_date").isNotNull()) &
    (F.col("amount").isNotNull()) &
    (F.col("amount") >= 0)
)
total+=1
assert valid_df.count() == 2, "NULL customer_id was not rejected"
passed+=1
print("Test 2 passed")
```

# Test-3

```python
test_orders = [
    (101, 1, "2026-09-20", 500.0),
    (None, 2, "2026-09-21", 1200.0),
    (103, 2, "2026-09-22", 800.0)
]

test_df = spark.createDataFrame(test_orders, columns)

valid_df = test_df.filter(
    (F.col("order_id").isNotNull()) &
    (F.col("customer_id").isNotNull()) &
    (F.col("order_date").isNotNull()) &
    (F.col("amount").isNotNull()) &
    (F.col("amount") >= 0)
)
total+=1
assert valid_df.count() == 2, "NULL order_id was not rejected"
passed+=1
print("Test 3 passed")
```

# Test-4 Valid data Should Remain

```python
test_orders = [
    (101, 1, "2026-09-20", 500.0),
    (102, 2, "2026-09-21", 1200.0),
    (103, 3, "2026-09-22", 800.0)
]

test_df = spark.createDataFrame(test_orders, columns)

valid_df = test_df.filter(
    (F.col("order_id").isNotNull()) &
    (F.col("customer_id").isNotNull()) &
    (F.col("order_date").isNotNull()) &
    (F.col("amount").isNotNull()) &
    (F.col("amount") >= 0)
)
total+=1
assert valid_df.count() == 3, "Valid records were incorrectly rejected"
passed+=1
print("Test 4 passed")
```

# Test-5 Duplicate order IDs

```python
test_orders = [
    (101, 1, "2026-09-20", 500.0),
    (101, 1, "2026-09-20", 500.0),
    (102, 2, "2026-09-21", 1200.0)
]

test_df = spark.createDataFrame(test_orders, columns)

dedup_df = test_df.dropDuplicates(["order_id"])

total+=1
assert dedup_df.count() == 2, "Duplicate order_id was not removed"
passed+=1
print("Test 5 passed")
```
